```python
# ============================================================
# EVALUATE LLaMA â€” Metrics + Submission + Checkpointing
# Resumes from last save if interrupted
# ============================================================
import subprocess
subprocess.run(['pip', 'install', 'textstat', 'sacrebleu', '-q'])
subprocess.run(['pip', 'install', 'git+https://github.com/feralvam/easse.git', '-q'])

import torch, numpy as np, time, ast, os, json, zipfile
import pandas as pd
from easse.sari import corpus_sari
from transformers import AutoTokenizer, AutoModel
import sacrebleu, textstat
import torch.nn.functional as F
from groq import Groq

device = 'cuda' if torch.cuda.is_available() else 'cpu'

# â”€â”€ Load val data â”€â”€
data_path = '/content/drive/MyDrive/SimpleText2025/cochrane-auto/data'
val_df    = pd.read_csv(f'{data_path}/cochraneauto_sents_val.csv')
val_df.dropna(subset=['complex', 'simple', 'label'], inplace=True)

def parse_simple(s):
    try:
        lst = ast.literal_eval(s)
        return ' '.join(lst) if isinstance(lst, list) else str(lst)
    except:
        return str(s)

val_df['simple_text'] = val_df['simple'].apply(parse_simple)
VALID = {'rephrase', 'delete', 'split', 'merge', 'copy'}
val_df = val_df[val_df['label'].str.strip().str.lower().isin(VALID)].reset_index(drop=True)
print(f"âœ… Val data loaded: {len(val_df)} sentences")

# â”€â”€ Groq client â”€â”€
os.environ["GROQ_API_KEY"] = "YOUR_GROQ_API_KEY"  # â† paste your key
client = Groq(api_key=os.environ["GROQ_API_KEY"])

# Verify key
try:
    resp = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[{"role": "user", "content": "Say READY"}],
        max_tokens=5
    )
    print(f"âœ… Groq key works: {resp.choices[0].message.content}")
except Exception as e:
    print(f"âŒ Groq key failed: {e}")
    raise

SYSTEM_PROMPT = """You are a medical text simplifier. Rewrite the sentence in plain English for a 13-year-old with no medical background.
RULES:
- Output ONLY the rewritten sentence, nothing else
- Always change the wording â€” never copy input unchanged
- Replace medical jargon with everyday words:
  * RCT â†’ clinical trial
  * odds ratio / OR â†’ chance of
  * heterogeneity â†’ variation between studies
  * glucocorticoids â†’ steroid medicines
  * confidence interval / CI â†’ range of uncertainty
  * I2 â†’ how different the study results were
- Keep sentences short and clear
- Target Grade 8 reading level"""

def simplify_with_llm(sentence, max_retries=3):
    for attempt in range(max_retries):
        try:
            resp = client.chat.completions.create(
                model="llama-3.1-8b-instant",
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user",   "content": sentence}
                ],
                max_tokens=150,
                temperature=0.3
            )
            result = resp.choices[0].message.content.strip()
            if result.strip() == sentence.strip():
                resp2 = client.chat.completions.create(
                    model="llama-3.1-8b-instant",
                    messages=[
                        {"role": "system", "content": SYSTEM_PROMPT},
                        {"role": "user",   "content": f"Rewrite in simple words for a 13-year-old. Do NOT copy:\n{sentence}"}
                    ],
                    max_tokens=150,
                    temperature=0.7
                )
                result = resp2.choices[0].message.content.strip()
            return result
        except Exception as e:
            print(f"  Retry {attempt+1}: {e}")
            time.sleep(5)
    return sentence

# â”€â”€ Checkpoint paths â”€â”€
CKPT_PATH    = '/content/drive/MyDrive/SimpleText2025/outputs/llama_val_checkpoint.json'
VAL_OUT_PATH = '/content/drive/MyDrive/SimpleText2025/outputs/tokatrons_task11_LLaMA8B_val.json'
VAL_ZIP_PATH = VAL_OUT_PATH.replace('.json', '.zip')
os.makedirs(os.path.dirname(VAL_OUT_PATH), exist_ok=True)

val_sentences = val_df['complex'].tolist()
val_refs      = val_df['simple_text'].tolist()
total         = len(val_sentences)

# â”€â”€ Load checkpoint if exists â”€â”€
if os.path.exists(CKPT_PATH):
    with open(CKPT_PATH) as f:
        ckpt = json.load(f)
    val_preds_llm = ckpt['predictions']
    start_from    = len(val_preds_llm)
    print(f"âœ… Resuming from sentence {start_from}/{total}")
else:
    val_preds_llm = []
    start_from    = 0
    print(f"Starting fresh â€” {total} sentences")

# â”€â”€ Generate predictions with checkpointing â”€â”€
start_time = time.time()

for i in range(start_from, total):
    sent = val_sentences[i]
    pred = simplify_with_llm(sent)
    val_preds_llm.append(pred)
    time.sleep(0.5)

    # Save checkpoint every 50 sentences
    if (i + 1) % 50 == 0:
        with open(CKPT_PATH, 'w') as f:
            json.dump({'predictions': val_preds_llm}, f)

        elapsed = time.time() - start_time
        done    = i + 1 - start_from
        eta     = (elapsed / done) * (total - i - 1) if done > 0 else 0
        changed = sum(1 for p, s in zip(val_preds_llm, val_sentences[:i+1])
                      if p.strip() != s.strip())
        print(f"  {i+1}/{total} | Changed: {changed} | ETA: {eta/60:.1f} min | Saved âœ…")

# Final checkpoint save
with open(CKPT_PATH, 'w') as f:
    json.dump({'predictions': val_preds_llm}, f)
print(f"\nâœ… All {total} predictions generated")

# â”€â”€ Build submission format â”€â”€
val_output = []
for i, (pred, sent) in enumerate(zip(val_preds_llm, val_sentences)):
    row = val_df.iloc[i]
    val_output.append({
        'pair_id':    str(row['pair_id'])  if 'pair_id'  in val_df.columns else f'val_{i}',
        'para_id':    int(row['para_id'])  if 'para_id'  in val_df.columns else 0,
        'sent_id':    int(row['sent_id'])  if 'sent_id'  in val_df.columns else i,
        'complex':    sent,
        'prediction': pred if pred.strip() != "" else sent,
        'run_id':     'tokatrons_task11_LLaMA8B_val'
    })

with open(VAL_OUT_PATH, 'w') as f:
    json.dump(val_output, f, indent=2)
with zipfile.ZipFile(VAL_ZIP_PATH, 'w') as zf:
    zf.write(VAL_OUT_PATH, arcname='tokatrons_task11_LLaMA8B_val.json')

print(f"âœ… Val submission saved: {VAL_ZIP_PATH}")
print(f"   Sample entry:\n{json.dumps(val_output[0], indent=4)}")

# â”€â”€ Filter empty for scoring â”€â”€
valid   = [(p, s, r) for p, s, r in zip(val_preds_llm, val_sentences, val_refs)
           if p.strip() != ""]
preds_c = [x[0] for x in valid]
src_c   = [x[1] for x in valid]
refs_c  = [x[2] for x in valid]
print(f"\nValid predictions: {len(preds_c)}/{len(val_preds_llm)}")

# â”€â”€ SARI â”€â”€
sari = corpus_sari(
    orig_sents=src_c,
    sys_sents=preds_c,
    refs_sents=[refs_c]
)
print(f"âœ… SARI: {sari:.4f}")

# â”€â”€ BLEU â”€â”€
bleu = sacrebleu.corpus_bleu(preds_c, [refs_c])
print(f"âœ… BLEU: {bleu.score:.4f}")

# â”€â”€ BERTScore (manual) â”€â”€
print("Computing BERTScore...")
bert_tok   = AutoTokenizer.from_pretrained('bert-base-uncased')
bert_model = AutoModel.from_pretrained('bert-base-uncased').to(device)
bert_model.eval()

def mean_pool(token_embeds, attention_mask):
    mask = attention_mask.unsqueeze(-1).expand(token_embeds.size()).float()
    return (token_embeds * mask).sum(1) / mask.sum(1).clamp(min=1e-9)

def get_embeddings(texts, batch_size=64):
    all_embs = []
    for i in range(0, len(texts), batch_size):
        batch = texts[i:i+batch_size]
        enc   = bert_tok(batch, return_tensors='pt', truncation=True,
                         padding=True, max_length=128).to(device)
        with torch.no_grad():
            out = bert_model(**enc)
        emb = mean_pool(out.last_hidden_state, enc['attention_mask'])
        all_embs.append(F.normalize(emb, dim=-1).cpu())
    return torch.cat(all_embs, dim=0)

pred_embs = get_embeddings(preds_c)
ref_embs  = get_embeddings(refs_c)
bert_f1   = (pred_embs * ref_embs).sum(dim=-1).mean().item()
print(f"âœ… BERTScore F1: {bert_f1:.4f}")

# â”€â”€ FKGL â”€â”€
src_fkgl  = np.mean([textstat.flesch_kincaid_grade(s) for s in src_c])
pred_fkgl = np.mean([textstat.flesch_kincaid_grade(p) for p in preds_c])
ref_fkgl  = np.mean([textstat.flesch_kincaid_grade(r) for r in refs_c])

# â”€â”€ Compression stats â”€â”€
src_lens   = [len(s.split()) for s in src_c]
pred_lens  = [len(p.split()) for p in preds_c]
comp_ratio = np.mean([p/s if s > 0 else 1.0 for p, s in zip(pred_lens, src_lens)])
deletions  = sum(1 for p, s in zip(pred_lens, src_lens) if p < s)
identical  = sum(1 for p, s in zip(preds_c, src_c) if p.strip() == s.strip())

# â”€â”€ Final table â”€â”€
print(f"\n{'='*62}")
print(f"METRIC COMPARISON â€” Val Set ({len(preds_c)} sentences)")
print(f"{'='*62}")
print(f"{'Metric':<25} {'BART v2':>10} {'LLaMA 8B':>10}  Context")
print(f"{'-'*62}")
print(f"{'SARI':<25} {'33.23':>10} {sari:>10.4f}  â†‘ higher better")
print(f"{'BLEU':<25} {'6.89':>10} {bleu.score:>10.4f}  â†‘ higher better")
print(f"{'BERTScore F1':<25} {'0.634':>10} {bert_f1:>10.4f}  â†‘ higher better")
print(f"{'-'*62}")
print(f"{'FKGL (source)':<25} {'13.03':>10} {src_fkgl:>10.2f}  input complexity")
print(f"{'FKGL (prediction)':<25} {'16.00':>10} {pred_fkgl:>10.2f}  â†“ lower = simpler")
print(f"{'FKGL (reference)':<25} {'8.07':>10} {ref_fkgl:>10.2f}  target ~8")
print(f"{'-'*62}")
print(f"{'Compression Ratio':<25} {'2.85':>10} {comp_ratio:>10.4f}  â†“ lower = shorter")
print(f"{'Deletion Proportion':<25} {'0.073':>10} {deletions/len(preds_c):>10.4f}  fraction made shorter")
print(f"{'Identical to Source':<25} {'0.005':>10} {identical/len(preds_c):>10.4f}  â†“ lower = more changed")
print(f"{'='*62}")

# â”€â”€ LaTeX row â”€â”€
print(f"\nLaTeX table row:")
print(f"LLaMA-3.1-8B zero-shot & {sari:.2f} & {bleu.score:.2f} & {bert_f1:.3f} & {pred_fkgl:.2f} \\\\")
print(f"\nVal submission: {VAL_ZIP_PATH}")

# â”€â”€ Clean up checkpoint after successful completion â”€â”€
if os.path.exists(CKPT_PATH):
    os.remove(CKPT_PATH)
    print("âœ… Checkpoint cleaned up")
```

```python
# ============================================================
# CELL 2 â€” TASK 1.2 SIMPLIFICATION FUNCTION
# Two-stage approach (best 2025 system used this):
#   Stage 1: Summarise the abstract
#   Stage 2: Simplify using summary as context guide
# ============================================================

SUMMARY_PROMPT = """You are a biomedical research assistant.
Read the following Cochrane systematic review abstract and write a 2-3 sentence plain-language summary capturing the main finding.
Output ONLY the summary, nothing else."""


SIMPLIFY_DOC_PROMPT = """You are a medical text simplifier. Simplify the following Cochrane abstract for a general audience (Grade 8 reading level).

RULES:
- Output ONLY the simplified text, sentence by sentence matching the original structure
- Replace ALL medical jargon with everyday words:
  * RCT/randomised controlled trial â†’ clinical trial
  * odds ratio / OR â†’ chance of
  * confidence interval / CI â†’ range of uncertainty
  * statistically significant â†’ clearly shown by the data
  * heterogeneity â†’ variation between studies
  * glucocorticoids â†’ steroid medicines
  * systematic review â†’ large study that combines many smaller studies
- Keep ALL factual information â€” numbers, percentages, study counts
- Make sentences shorter and clearer
- Do NOT add information not in the original
- Target reading level: Grade 8"""

def summarise_abstract(abstract, max_retries=3):
    for attempt in range(max_retries):
        try:
            resp = client.chat.completions.create(
                model="llama-3.1-8b-instant",
                messages=[
                    {"role": "system", "content": SUMMARY_PROMPT},
                    {"role": "user",   "content": abstract}
                ],
                max_tokens=150,
                temperature=0.3
            )
            return resp.choices[0].message.content.strip()
        except Exception as e:
            print(f"  Summary retry {attempt+1}: {e}")
            time.sleep(5)
    return ""

def simplify_document(abstract, summary, max_retries=3):
    prompt = f"""Here is a plain-language summary of the abstract to guide you:
SUMMARY: {summary}

Now simplify this full abstract:
{abstract}"""

    for attempt in range(max_retries):
        try:
            resp = client.chat.completions.create(
                model="llama-3.1-8b-instant",
                messages=[
                    {"role": "system", "content": SIMPLIFY_DOC_PROMPT},
                    {"role": "user",   "content": prompt}
                ],
                max_tokens=600,   # documents are longer than sentences
                temperature=0.3
            )
            return resp.choices[0].message.content.strip()
        except Exception as e:
            print(f"  Simplify retry {attempt+1}: {e}")
            time.sleep(5)
    return abstract  # fallback

# Test on one document
print("=== TEST ON 1 DOCUMENT ===")
test_doc = test12[0]['complex']
print(f"INPUT ({len(test_doc.split())} words):")
print(test_doc[:300])

summary = summarise_abstract(test_doc)
print(f"\nSUMMARY: {summary}")

simplified = simplify_document(test_doc, summary)
print(f"\nSIMPLIFIED ({len(simplified.split())} words):")
print(simplified[:500])

orig_words = len(test_doc.split())
simp_words = len(simplified.split())
print(f"\nCompression: {orig_words} â†’ {simp_words} words ({simp_words/orig_words:.2f}x)")
```

```python
# ============================================================
# CELL 3 FAST â€” Single API call per document
# 666 docs Ã— 1 call Ã— ~1s = ~11 min
# ============================================================

CKPT_PATH = f'{base}/outputs/llama_task12_checkpoint.json'
OUT_PATH  = f'{base}/outputs/tokatrons_task12_LLaMA8B.json'
ZIP_PATH  = OUT_PATH.replace('.json', '.zip')
os.makedirs(os.path.dirname(OUT_PATH), exist_ok=True)

# Single-stage prompt â€” no summary step needed
SIMPLIFY_DOC_PROMPT = """You are a medical text simplifier. Simplify this Cochrane abstract for a general audience.

RULES:
- Output ONLY the simplified text, nothing else
- Keep the same paragraph structure as the original
- Replace ALL medical jargon with everyday words:
  * RCT â†’ clinical trial
  * odds ratio / OR â†’ chance of
  * confidence interval / CI â†’ range of uncertainty
  * statistically significant â†’ clearly shown by the data
  * heterogeneity â†’ variation between studies
  * glucocorticoids â†’ steroid medicines
  * systematic review â†’ large study combining many smaller studies
- Keep ALL numbers and factual information
- Make sentences shorter and clearer
- Do NOT add information not in the original
- Target Grade 8 reading level"""

def simplify_document_fast(abstract, max_retries=3):
    for attempt in range(max_retries):
        try:
            resp = client.chat.completions.create(
                model="llama-3.1-8b-instant",
                messages=[
                    {"role": "system", "content": SIMPLIFY_DOC_PROMPT},
                    {"role": "user",   "content": abstract}
                ],
                max_tokens=800,
                temperature=0.3
            )
            return resp.choices[0].message.content.strip()
        except Exception as e:
            print(f"  Retry {attempt+1}: {e}")
            time.sleep(5)
    return abstract

# Test on 1 doc first
print("=== QUICK TEST ===")
test_doc   = test12[0]['complex']
simplified = simplify_document_fast(test_doc)
print(f"IN  ({len(test_doc.split())} words): {test_doc[:150]}")
print(f"OUT ({len(simplified.split())} words): {simplified[:150]}")
print(f"Compression: {len(simplified.split())/len(test_doc.split()):.2f}x")
print()

total = len(test12)

# Load checkpoint
if os.path.exists(CKPT_PATH):
    with open(CKPT_PATH) as f:
        ckpt = json.load(f)
    results    = ckpt['results']
    start_from = len(results)
    print(f"âœ… Resuming from {start_from}/{total}")
else:
    results    = []
    start_from = 0
    print(f"Starting fresh â€” {total} documents")

print(f"Estimated time: {(total - start_from) * 1.0 / 60:.0f} min\n")

start_time = time.time()

for i in range(start_from, total):
    doc      = test12[i]
    pred     = simplify_document_fast(doc['complex'])
    results.append({
        'pair_id':    doc['pair_id'],
        'source':     doc['source'],
        'complex':    doc['complex'],
        'prediction': pred,
        'run_id':     'tokatrons_task12_LLaMA8B'
    })
    time.sleep(0.5)

    if (i + 1) % 50 == 0:
        with open(CKPT_PATH, 'w') as f:
            json.dump({'results': results}, f)
        elapsed = time.time() - start_time
        done    = i + 1 - start_from
        eta     = (elapsed / done) * (total - i - 1)
        print(f"  {i+1}/{total} | ETA: {eta/60:.1f} min | Saved âœ…")

# Save
with open(OUT_PATH, 'w') as f:
    json.dump(results, f, indent=2)
with zipfile.ZipFile(ZIP_PATH, 'w') as zf:
    zf.write(OUT_PATH, arcname='tokatrons_task12_LLaMA8B.json')

changed = sum(1 for r in results if r['prediction'] != r['complex'])
empty   = sum(1 for r in results if r['prediction'].strip() == "")
print(f"\nâœ… Done! Total: {len(results)} | Changed: {changed} | Empty: {empty}")
print(f"âœ… Submit: {ZIP_PATH}")

if os.path.exists(CKPT_PATH):
    os.remove(CKPT_PATH)
```

```python
# ============================================================
# BART v2 LOCAL â€” full 48809 sentences, ~60 min, no API limits
# ============================================================
import subprocess
subprocess.run(['pip', 'install', 'transformers', 'sentencepiece', '-q'])

import os, json, zipfile, time, torch
from transformers import BartTokenizer, BartForConditionalGeneration
from google.colab import drive
drive.mount('/content/drive')

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Device: {device}")

# Load BART v2
v2_path   = '/content/drive/MyDrive/SimpleText2025/models/bart-plan-guided'
tokenizer = BartTokenizer.from_pretrained(v2_path)
model     = BartForConditionalGeneration.from_pretrained(v2_path).to(device)

model.generation_config.forced_bos_token_id = 0

special_tokens = ['[REPHRASE]','[DELETE]','[SPLIT]','[MERGE]','[COPY]']
tokenizer.add_tokens(special_tokens)
model.resize_token_embeddings(len(tokenizer))
model.eval()
print(f"âœ… BART v2 loaded")

# Paths
DATA_PATH = '/content/drive/MyDrive/input/sentences_src.json'
OUT_DIR   = '/content/drive/MyDrive/SimpleText2025/outputs'
OUT_PATH  = f'{OUT_DIR}/tokatrons_task11_BART_2026.json'
ZIP_PATH  = OUT_PATH.replace('.json', '.zip')
CKPT_PATH = f'{OUT_DIR}/bart_2026_checkpoint.json'
os.makedirs(OUT_DIR, exist_ok=True)

with open(DATA_PATH) as f:
    src_data = json.load(f)
total     = len(src_data)
sentences = [d['complex'] for d in src_data]
print(f"âœ… Loaded {total} sentences")

# â”€â”€ Load checkpoint â”€â”€
if os.path.exists(CKPT_PATH):
    with open(CKPT_PATH) as f:
        ckpt = json.load(f)
    preds      = ckpt['preds']
    start_from = len(preds)
    print(f"âœ… Resuming from {start_from}/{total}")
    print(f"   Delete {CKPT_PATH} to start fresh")
else:
    preds      = []
    start_from = 0
    print("Starting fresh")

BATCH_SIZE    = 16        # â† smaller batch so checkpoint triggers more often
start_time    = time.time()
last_log_time = time.time()
LOG_INTERVAL  = 10        # log every 10 seconds
CKPT_EVERY    = 100       # checkpoint every 100 sentences

print(f"\nðŸš€ Starting from sentence {start_from}, {total - start_from} remaining...\n")

for i in range(start_from, total, BATCH_SIZE):
    batch   = sentences[i:i+BATCH_SIZE]
    guided  = ['[REPHRASE] ' + s for s in batch]
    enc     = tokenizer(guided, return_tensors='pt', max_length=256,
                        truncation=True, padding=True).to(device)
    with torch.no_grad():
        out = model.generate(
            **enc,
            max_length=128,
            num_beams=2,
            forced_bos_token_id=0
        )
    decoded = tokenizer.batch_decode(out, skip_special_tokens=True)
    preds.extend(decoded)

    now     = time.time()
    elapsed = now - start_time
    done    = len(preds) - start_from
    speed   = done / elapsed if elapsed > 0 else 0
    eta     = (total - len(preds)) / speed if speed > 0 else 0

    # â”€â”€ Log every 10 seconds â”€â”€
    if now - last_log_time >= LOG_INTERVAL:
        print(f"  [{time.strftime('%H:%M:%S')}] {len(preds)}/{total} | "
              f"{done/(elapsed/60):.0f} sent/min | "
              f"Elapsed: {elapsed/60:.1f} min | "
              f"ETA: {eta/60:.1f} min")
        last_log_time = now

    # â”€â”€ Checkpoint every 100 sentences â”€â”€
    if len(preds) % CKPT_EVERY < BATCH_SIZE:
        with open(CKPT_PATH, 'w') as f:
            json.dump({'preds': preds}, f)
        print(f"  ðŸ’¾ Checkpoint saved at {len(preds)}/{total} "
              f"[{time.strftime('%H:%M:%S')}]")

# â”€â”€ Final save â”€â”€
with open(CKPT_PATH, 'w') as f:
    json.dump({'preds': preds}, f)

elapsed_total = time.time() - start_time
print(f"\nâœ… All {len(preds)} predictions done in {elapsed_total/60:.1f} min")

# â”€â”€ Build submission â”€â”€
submission = []
for i, (entry, pred) in enumerate(zip(src_data, preds)):
    submission.append({
        "pair_id":    str(entry['pair_id']),
        "source":     entry.get('source', 'Cochrane-auto 2026'),
        "language":   entry.get('language', 'en'),
        "para_id":    int(entry['para_id']),
        "sent_id":    int(entry['sent_id']),
        "complex":    entry['complex'],
        "prediction": pred if pred.strip() != "" else entry['complex'],
        "run_id":     "tokatrons_task11_BART_2026"
    })

with open(OUT_PATH, 'w') as f:
    json.dump(submission, f, indent=2)
with zipfile.ZipFile(ZIP_PATH, 'w') as zf:
    zf.write(OUT_PATH, arcname='tokatrons_task11_BART_2026.json')

identical = sum(1 for s in submission
                if s['prediction'].strip() == s['complex'].strip())
print(f"\nâœ… Submission saved: {ZIP_PATH}")
print(f"ðŸ“Š Total: {len(submission)} | Changed: {len(submission)-identical} | Identical: {identical}")
print(f"â±  Total time: {elapsed_total/60:.1f} min")
print(f"\nSample:")
for entry in submission[:2]:
    print(json.dumps(entry, indent=4))

if os.path.exists(CKPT_PATH):
    os.remove(CKPT_PATH)
    print("\nâœ… Checkpoint cleaned up")
```
